## **Data Cleansing and normalization**

These steps are required for a data to be made ready for Machine Learning

*Data Cleansing:* Is to fix or remove incorrect, incomplete or messy data

*Data normalization:* This (also called scaling) is done to normalize the data so that no particular feature dominates.

```python
import pandas as pd

data1 = {"Name" :["Deepak", "Param", "Sanjiv","David", "Jatin", None, "Sanjiv"],
         "Age" :[ 25, None, 30.2, 28, 52, 110, 30],
         "Salary" :[50000, 20000, None, 30000, 55000, 80000, None],
         "City" : ["Delhi", "delhi", "Mumbai", "Bangalore", "Bomday", "Pune", "Mumbai"]}

df3=pd.DataFrame(data1)
print(df3)

df31=df3
print(df31)
```

### **Data Cleansing**

Problems with data are :
- Missing values : some cells are empty
- Duplicates : some rows are repeated
- Wrong format : say "4" is written instead of 4.
- Outliers : For example Age is given to be 1000
- inconsistent : for example in city Delhi, DElhi, delhi are written.


The option **inplace** (by default **False**) is made to be **True** then it removes the desired things from the database.

```python
df3.info()
```

```python
df3.isnull().sum()         # Counts Null per column

df3.dropna()               # Drops rows with null entries

df3.dropna(subset=["Name", "Salary"])  # Drops the rows if specific columns have null entries
```

```python
df3.fillna(0)       # Fill 0 where there is missing or null values
```

```python
df3["Age"].fillna(df3["Age"].mean())
```

```python
df31.drop_duplicates(inplace=True) # The duplicate rows will be deleted in df31
df31
```

```python
df31.dropna(inplace=True)                  # This won't change the df31 whereas if "inplace = True" the rows with wrong entries will be deleted in df31
```

```python
# Fixing data-type

df31["Age"] = df31["Age"].astype(int)      #
```

```python
# Standardizing text

df31["City"] = df31["City"].str.upper()
```
